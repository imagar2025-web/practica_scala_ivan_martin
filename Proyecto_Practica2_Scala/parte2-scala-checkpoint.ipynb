```scala
## Ejercicio 4 — Función para determinar si una mano se pasa de 21
**Solución :** Creación de una función pura que evalúa un entero y devuelve un booleano.
```

```scala
def bust(mano: Int): Boolean = {
  mano > 21
}

println(s"18: ${bust(18)}")
println(s"21: ${bust(21)}")
println(s"22: ${bust(22)}")
println(s"30: ${bust(30)}")
```

```scala
## Ejercicio 5 — Comparación de dos manos
**Solución:** Función iterativa simple usando condicionales.

**Documentación:**
* **Empate:** Si ambas manos tienen el mismo valor, la condición `h1 > h2` es falsa, por lo que entra en el `else` y devuelve `h2`. El comportamiento es correcto porque al ser iguales, devolver cualquiera de los dos es válido, pero a nivel lógico se podría modificar para devolver un mensaje o manejar un estado de "empate".
```

```scala
def maxHand(h1: Int, h2: Int): Int = {
  if (h1 > h2) h1 else h2
}

println(s"17 y 19 -> Mayor: ${maxHand(17, 19)}")
println(s"20 y 18 -> Mayor: ${maxHand(20, 18)}")
println(s"21 y 21 -> Mayor: ${maxHand(21, 21)}")
```

```scala
## Ejercicio 6 — Decidir el ganador de una partida
**Solución:** Uso de la función `bust` anterior dentro de una nueva función para aplicar reglas complejas.

**Documentación de las pruebas:**
* **26 y 20:** handA supera 21, devuelve handB (20). (Condición 2)
* **18 y 22:** handB supera 21, devuelve handA (18). (Condición 3)
* **24 y 25:** Ambas superan 21, devuelve 0. (Condición 1)
* **17 y 19:** Ninguna supera, devuelve la mayor (19). (Condición 4 - else)
* **21 y 20:** Ninguna supera, devuelve la mayor (21). (Condición 4 - else)
```

```scala
def ganador(handA: Int, handB: Int): Int = {
  if (bust(handA) && bust(handB)) 0
  else if (bust(handA)) handB
  else if (bust(handB)) handA
  else maxHand(handA, handB)
}

println(s"26 y 20: ${ganador(26, 20)}")
println(s"18 y 22: ${ganador(18, 22)}")
println(s"24 y 25: ${ganador(24, 25)}")
println(s"17 y 19: ${ganador(17, 19)}")
println(s"21 y 20: ${ganador(21, 20)}")
```

```scala
## Ejercicio 7 — Arrays y mutabilidad
**Solución:** Manipulación de posiciones en un Array.

**Documentación:**
* **Por qué cambian los elementos:** En Scala, `Array` es una estructura de datos mutable por naturaleza. Declararlo como `val` solo impide reasignar la variable `jugadores` a un *nuevo* Array, pero no impide cambiar el contenido interno del Array existente.
* **Error del valor 500:** Scala es fuertemente tipado. El Array se infirió como `Array[String]`, por lo que no acepta tipos `Int`.
```

```scala
val jugadores = Array("Alex", "Chen", "Marta")
println(s"Original: ${jugadores.mkString(", ")}")

jugadores(0) = "Sindhu"
println(s"Modificado: ${jugadores.mkString(", ")}")

// Esto dará error intencionadamente
jugadores(1) = 500
```

```scala
## Ejercicio 8 — Creación e inicialización de Arrays
**Solución planteada:** Instanciación manual de Arrays con tamaño predefinido.

**Pregunta:** ¿Qué valores contiene un `Array[Int]` recién creado antes de asignar manualmente sus elementos?
* **Respuesta:** Contiene el valor por defecto del tipo de dato, que en el caso de `Int` es `0`.
```

```scala
## Ejercicio 9 — Recorrer un Array con while
**Solución:** Uso de bucle `while` clásico y contador para evaluar las manos.

**Documentación:**
* **Contador mutable:** `i` debe ser `var` porque su valor tiene que cambiar (incrementarse) en cada iteración del bucle.
* **Qué ocurriría sin incremento:** Se produciría un bucle infinito, ya que `i` siempre sería 0 y la condición `i < manosArray.length` siempre sería verdadera.
* **Estilo imperativo:** El uso de la variable `var i` y el bucle `while` manejando explícitamente el estado del contador son características fundamentales del estilo imperativo.
```

```scala
val manosArray = Array(17, 24, 21, 19, 26)
var i = 0

while (i < manosArray.length) {
  val sePasa = manosArray(i) > 21
  println(s"Mano: ${manosArray(i)} - ¿Se pasa?: $sePasa")
  i += 1
}
```

```scala
## Ejercicio 10 — Listas e inmutabilidad
**Solución :** Operaciones básicas con listas inmutables (`List`).

**Documentación:**
* **Por qué `::` produce una lista nueva:** Las listas (`List`) en Scala son inmutables por defecto. Cualquier operación que aparentemente "modifique" una lista en realidad genera y devuelve una nueva lista en memoria, preservando la original.
```

```scala
val jugadoresList = List("Alex", "Chen", "Marta")
println(s"Lista original: $jugadoresList")

val nuevaLista = "Sindhu" :: jugadoresList
println(s"Lista nueva: $nuevaLista")
println(s"Lista original sigue igual: $jugadoresList")

println(s"Longitud: ${jugadoresList.length}")
println(s"Invertida: ${jugadoresList.reverse}")
```

```scala
## Ejercicio 12 — Operadores relacionales y lógicos
**Solución planteada:** Evaluación de expresiones booleanas.

**Documentación:**

| Expresión | Resultado | Explicación |
| :--- | :--- | :--- |
| `handA > handB` | `false` | 18 no es mayor que 21. |
| `handB == 21` | `true` | handB vale exactamente 21. |
| `handC != 21` | `true` | handC (25) es distinto de 21. |
| `handA <= 21 && handB <= 21` | `true` | Ambas condiciones se cumplen. |
| `handA > 21 || handC > 21` | `true` | handC supera 21, así que el operador OR da verdadero. |
| `!(handB > 21)` | `true` | handB no es mayor a 21, por lo que el negador (!) invierte el false a true. |
```

```scala
val handA = 18
val handB = 21
val handC = 25

println(s"¿Es handA mayor que handB?: ${handA > handB}")
println(s"¿Es handB igual a 21?: ${handB == 21}")
println(s"¿Es handC diferente de 21?: ${handC != 21}")
println(s"¿Son handA y handB menores o iguales a 21?: ${handA <= 21 && handB <= 21}")
println(s"¿Alguna supera 21?: ${handA > 21 || handC > 21}")
println(s"¿No se ha pasado handB de 21?: ${!(handB > 21)}")
```

```scala
## Ejercicio 13 — foreach y funciones como valores
**Solución:** Comparativa entre iteración manual (imperativa) e iteración funcional.

**Documentación:**
* **Cuál necesita contador:** Versión A (bucle `while`).
* **Cuál necesita variable var:** Versión A (`var j = 0`).
* **Estilo funcional:** Versión B se aproxima al estilo funcional porque usa abstracciones de orden superior (`foreach`), evita estados mutables (no usa `var`) y se enfoca en "qué" hacer en lugar de "cómo" iterar.
```

```scala
val manosE13 = Array(17, 24, 21, 26, 18)

// Versión A
println("--- Versión A (while) ---")
var j = 0
while (j < manosE13.length) {
  val sePasa = manosE13(j) > 21
  println(s"Mano: ${manosE13(j)} - Bust: $sePasa")
  j += 1
}

// Versión B
println("--- Versión B (foreach) ---")
manosE13.foreach(m => println(s"Mano: $m - Bust: ${m > 21}"))
```

```scala
## Ejercicio 14 — Efectos secundarios y estilo de programación
**Solución:** Funciones puras vs impuras.

**Documentación:**
* La función `sumarAlTotal` modifica una variable fuera de su ámbito local porque interactúa con variables externas de nivel superior (`var total`).
* **Efecto secundario:** La alteración de la variable global `total` es el efecto secundario de llamar a la función.

| Característica | sumarAlTotal | sumar |
| :--- | :--- | :--- |
| Modifica datos externos | Sí | No |
| Devuelve un resultado calculado | No (Unit) | Sí (Int) |
| Utiliza efecto secundario | Sí | No |
| Estilo predominante | Imperativo | Funcional |
```

```scala
var total = 0
def sumarAlTotal(valor: Int) = {
  total = total + valor
}

sumarAlTotal(5)
println(s"Total tras llamada 1: $total")
sumarAlTotal(10)
println(s"Total tras llamada 2: $total")

// Segunda versión (Pura)
def sumar(a: Int, b: Int): Int = {
  a + b
}
println(s"Llamada pura: ${sumar(5, 10)}")
```

```scala
## Ejercicio 15 — Programa integrado: torneo de Twenty-One
**Solución:** Integración de listas, arrays, bucles `while`, condicionales y lógica relacional.
```

```scala
val jugadoresTorneo = List("Alex", "Chen", "Marta", "Sindhu")
val manosTorneo = Array(18, 24, 21, 20)

def procesarRonda(manos: Array[Int]): Unit = {
  var mejorPuntuacion = 0
  var k = 0
  
  println("--- Resultados de la ronda ---")
  while (k < manos.length) {
    val jugador = jugadoresTorneo(k)
    val mano = manos(k)
    
    // Comprobación directa (sin función bust)
    if (mano > 21) {
      println(s"$jugador: Se ha pasado con $mano")
    } else {
      println(s"$jugador: Válida con $mano")
      
      // Actualizar la mejor puntuación
      if (mano > mejorPuntuacion) {
        mejorPuntuacion = mano
      }
    }
    k += 1
  }
  println(s"-> La mejor puntuación válida es: $mejorPuntuacion\n")
}

// Ronda 1
println("RONDA 1")
procesarRonda(manosTorneo)

// Ampliación: Ronda 2
println("RONDA 2")
val manosRonda2 = Array(22, 19, 20, 21)
procesarRonda(manosRonda2)
```
